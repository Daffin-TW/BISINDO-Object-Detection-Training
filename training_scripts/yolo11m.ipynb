```python
from ultralytics import YOLO


MODEL = 'yolo11m'

yaml_path = os.path.join(data_path, 'data.yaml')

# Load model pretrained
model = YOLO(MODEL + '.pt')

# Train
model.train(
    data=yaml_path,
    epochs=30,
    imgsz=640,
    batch=16,
    patience=10,
    name=MODEL + '_train_val_set',
    device=0
)
```

```python
results = model.val(
    data=yaml_path,
    split='test',
    name=MODEL + '_test_set',
    save_json=True
)
```
